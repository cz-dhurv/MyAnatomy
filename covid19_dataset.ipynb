```python
# Install the required dependency (adding '!' for Colab)
!pip install kagglehub[pandas-datasets]

import kagglehub
from kagglehub import KaggleDatasetAdapter

# Set the path to the specific CSV file you want to load from the dataset
file_path = "covid_19_clean_complete.csv" # Example filename, replace with the exact file you need

# Load the latest version into a pandas DataFrame
df = kagglehub.load_dataset(
  KaggleDatasetAdapter.PANDAS,
  "imdevskp/corona-virus-report",
  file_path,
)

print("First 5 records:", df.head())
```

```python
df["Date"] = pd.to_datetime(df["Date"])
df
```

```python
df['Province/State'].fillna('Unknown', inplace=True)
df.head()
```

```python
state_data = df.groupby("Province/State")[["Confirmed","Deaths","Recovered","Active"]].sum().reset_index()
```

```python
state_data
```

```python
state_data['Recovery Rate (%)'] = (state_data['Recovered'] / state_data['Confirmed'] * 100).fillna(0).round(2)

display(state_data.sort_values(by='Recovery Rate (%)', ascending=False).head(10))
```

```python
state_data['Death Rate (%)'] = (state_data['Deaths'] / state_data['Confirmed'] * 100).fillna(0).round(2)

display(state_data.sort_values(by='Death Rate (%)', ascending=False).head(10))
```

```python
state_data.head()
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

plt.figure(figsize=(10, 6))
sns.scatterplot(data=state_data, x='Recovery Rate (%)', y='Death Rate (%)', alpha=0.7, color='purple')

plt.title('COVID-19: Death Rate vs Recovery Rate by Province/State', fontsize=14)
plt.xlabel('Recovery Rate (%)', fontsize=12)
plt.ylabel('Death Rate (%)', fontsize=12)
plt.grid(True, linestyle='--', alpha=0.5)

plt.show()
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

# Sort the data by Death Rate (%) to show the top provinces/states
top_10_death_rate = state_data.sort_values(by='Death Rate (%)', ascending=False).head(10)

plt.figure(figsize=(12, 7))
sns.barplot(data=top_10_death_rate, x='Death Rate (%)', y='Province/State', hue='Province/State', palette='viridis', legend=False)

plt.title('COVID-19: Top 10 Provinces/States by Death Rate (%)', fontsize=16)
plt.xlabel('Death Rate (%)', fontsize=12)
plt.ylabel('Province/State', fontsize=12)
plt.grid(axis='x', linestyle='--', alpha=0.5)
plt.tight_layout()
plt.show()
```
